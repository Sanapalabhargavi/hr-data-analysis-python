# 👥 IBM HR Analytics – Employee Attrition Analysis

This notebook explores employee attrition, compensation, job roles, satisfaction, experience, and workforce characteristics using Python.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', None)
```

## 2. Load Dataset

```python
# Update this path if your CSV has a different name or location.
file_path = '/content/IBM HR Analytics Employee.csv'
df = pd.read_csv(file_path)
df.head()
```

## 3. Data Understanding

```python
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
display(df.head())
display(df.dtypes)
```

## 4. Data Quality Checks

```python
print('Missing values:')
display(df.isnull().sum().sort_values(ascending=False).head(10))
print('Duplicate rows:', df.duplicated().sum())
```

## 5. Overall Attrition Analysis

```python
attrition_counts = df['Attrition'].value_counts()
attrition_rate = df['Attrition'].eq('Yes').mean() * 100
print(f'Overall attrition rate: {attrition_rate:.2f}%')
display(attrition_counts)
```

```python
attrition_counts.plot(kind='bar', title='Employee Attrition')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.show()
```

## 6. Attrition by Department and Job Role

```python
department_attrition = pd.crosstab(df['Department'], df['Attrition'], normalize='index') * 100
display(department_attrition)
jobrole_attrition = pd.crosstab(df['JobRole'], df['Attrition'], normalize='index') * 100
display(jobrole_attrition.sort_values('Yes', ascending=False))
```

## 7. Overtime and Attrition

```python
overtime_attrition = pd.crosstab(df['OverTime'], df['Attrition'], normalize='index') * 100
display(overtime_attrition)
```

## 8. Compensation Analysis

```python
print('Average Monthly Income:', round(df['MonthlyIncome'].mean(), 2))
income_by_role = df.groupby('JobRole')['MonthlyIncome'].agg(['count', 'mean', 'median']).sort_values('mean', ascending=False)
display(income_by_role)
```

## 9. Employee Satisfaction Analysis

```python
satisfaction_columns = ['JobSatisfaction', 'EnvironmentSatisfaction', 'RelationshipSatisfaction', 'WorkLifeBalance']
display(df.groupby('Attrition')[satisfaction_columns].mean().round(2))
```

## 10. Experience and Tenure Analysis

```python
tenure_columns = ['TotalWorkingYears', 'YearsAtCompany', 'YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager']
display(df.groupby('Attrition')[tenure_columns].mean().round(2))
```
