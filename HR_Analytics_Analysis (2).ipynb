# 👥 IBM HR Analytics – Employee Attrition Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', None)
```

```python
# Update this path if your CSV has a different name or location.
file_path = '/content/IBM HR Analytics Employee.csv'
df = pd.read_csv(file_path)
df.head()
```

```python
# 😊 Job Satisfaction and Attrition

satisfaction_attrition = pd.crosstab(
    df['JobSatisfaction'],
    df['Attrition'],
    normalize='index'
) * 100

display(satisfaction_attrition)
```

```python
# 📊 Job Satisfaction vs Attrition

satisfaction_attrition['Yes'].plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Attrition Rate by Job Satisfaction')
plt.xlabel('Job Satisfaction Level')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# 💰 Average Monthly Income by Attrition

income_attrition = df.groupby('Attrition')['MonthlyIncome'].mean()

display(income_attrition)
```

```python
# 📈 Experience and Attrition

experience_attrition = df.groupby('Attrition')[
    ['TotalWorkingYears', 'YearsAtCompany', 'YearsInCurrentRole',
     'YearsSinceLastPromotion', 'YearsWithCurrManager']
].mean().round(2)

display(experience_attrition)
```

```python
# 📊 Years at Company vs Attrition

tenure_attrition = df.groupby('Attrition')['YearsAtCompany'].mean()

tenure_attrition.plot(
    kind='bar',
    figsize=(7, 5)
)

plt.title('Average Years at Company by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Average Years at Company')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

The HR Analytics project provides a data-driven view of employee attrition and workforce characteristics.

The analysis found that the overall employee attrition rate is **16.12%**. Overtime is an important factor associated with attrition, with employees working overtime showing a substantially higher attrition rate than employees who do not work overtime.

These findings suggest that organizations can improve employee retention by managing excessive workloads, improving employee satisfaction, supporting career development, and focusing on employees in the early stages of their organizational tenure.
